```python
data = df_data.copy()
for column in ["Order Date", "Ship Date"]:
    data[column] = pd.to_datetime(data[column], format="%m/%d/%Y", errors="raise")

print("Order period:", data["Order Date"].min().date(), "to", data["Order Date"].max().date())

checks = pd.Series({
    "Ship dates before order dates": (data["Ship Date"] < data["Order Date"]).sum(),
    "Non-positive sales": (data["Sales"] <= 0).sum(),
    "Non-positive quantities": (data["Quantity"] <= 0).sum(),
    "Discounts outside 0 to 1": (~data["Discount"].between(0, 1)).sum(),
    "Nagetive Profit": (data["Profit"] < 0).sum(), #
})
display(checks.to_frame("Count"))
```

```python
# The baseline KPIs

total_sales = data["Sales"].sum()
total_profit = data["Profit"].sum()
order_counts = data["Order ID"].nunique()
unit_sold = data["Quantity"].sum()
print(f"The total slaes = R {total_sales:,.2f}")
print(f"The total profit = R {total_profit:,.2f}")
print(f"Profit margin = {(total_profit / total_sales):.2%}")
print(f"Order counts: {order_counts:,}")
print(f"Average order value = {total_sales / order_counts:.2f}")
print(f"Units Sold: {unit_sold:,}")
```

```python
# Monthly revenue trends 2014 to 2017
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%Y-%m-%d"
)
sales_monthly = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
    .reset_index(name="Revenue")
)

sales_monthly.head(10)
```

```python
import matplotlib.pyplot as plt

# Monthly revenue trends 2014 to 2017
sales_monthly = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

plt.figure(figsize=(11, 5))

sales_monthly.plot(
    kind="line",
    color="steelblue",
    linewidth=2,
    marker="o",
    markersize=3
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Top-selling products

top_prod = (
    df.groupby("Product Name")["Sales"].sum().sort_values(ascending = False)
    .reset_index(name="Revenue")
)

print("Top 10 products:")
top_prod.head(10)
```

```python

# Top 10 products by revenue

top_10 = (
    top_prod.nlargest(10, "Revenue")
    .sort_values("Revenue")
)

plt.figure(figsize=(14, 6))
plt.barh(
    top_10["Product Name"],
    top_10["Revenue"]
)
plt.bar_label(bars, fmt="{:,.0f}", padding=3)
plt.title("Top 10 Products by Revenue", fontsize=12)
plt.xlabel("Revenue")
plt.ylabel("Product Name")
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# High-value categories analysis

category_sales = (
    df.groupby("Category")["Sales"].sum().sort_values(ascending = False)
)
    
categoty = category_sales.plot(figsize = (10, 6), kind = 'bar')

for bars in categoty.containers:
    categoty.bar_label(bars, fmt="{:,.0f}", padding=3)

plt.title("Sales by category")
plt.xlabel("Categories")
plt.ylabel("Revenue")
plt.grid(axis = 'y', linestyle = '--', alpha = 0.5)
plt.show()
```

```python
# Regional sales performance

regional_sales = (
    df.groupby("Region")
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
    .sort_values("Revenue", ascending=False)
    .reset_index()
)

# Calculate profit margin per region
regional_sales["Profit Margin (%)"] = (
    regional_sales["Profit"] /
    regional_sales["Revenue"] * 100
)

print("Regional Sales Performance:")
display(regional_sales.round(2))
```

```python
# Region by sales analysis

region_sales = (
    df.groupby("Region")["Sales"].sum().sort_values(ascending = False)
)

region = region_sales.plot(figsize = (10, 6), kind = 'bar')

for bars in region.containers:
    region.bar_label(bars, fmt="{:,.0f}", padding=3)

plt.title("Sales by Region")
plt.xlabel("Regions")
plt.ylabel("Revenue")
plt.grid(axis = 'y', linestyle = '--', alpha = 0.5)
plt.show()
```

```python
# Regional revenue and profit visualization

region = regional_sales.plot(
    x="Region",
    y=["Revenue", "Profit"],
    kind="bar",
    figsize=(10, 6),
    color=["steelblue", "darkgreen"],
    width=0.7
)
for bars in region.containers:
    region.bar_label(bars, fmt="{:,.0f}", padding=3)
plt.title("Revenue and Profit by Region", fontsize=14)
plt.xlabel("Region")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.legend(title="Performance")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
